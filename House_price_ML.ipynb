```python
new_df.drop(columns=["sqft_lot"],inplace=True)
```

```python
Q1=new_df["sqft_living"].quantile(0.25)
Q3=new_df["sqft_living"].quantile(0.75)

IQR=Q3-Q1

upper_limit=Q3+(1.5*IQR)
lower_limit=Q1-(1.5*IQR)

new_df["sqft_living"]=np.where(
    new_df["sqft_living"]>upper_limit,
    upper_limit,
    np.where(
        new_df["sqft_living"]<lower_limit,
        lower_limit,
        new_df["sqft_living"]
    )
)
```

```python
Q1=new_df["bathrooms"].quantile(0.25)
Q3=new_df["bathrooms"].quantile(0.75)

IQR=Q3-Q1

upper_limit1=Q3+(1.5*IQR)
lower_limit1=Q1-(1.5*IQR)

new_df["bathrooms"]=np.where(
    new_df["bathrooms"]>upper_limit1,
    upper_limit1,
    np.where(
        new_df["bathrooms"]<lower_limit1,
        lower_limit1,
        new_df["bathrooms"]
    )
)
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(new_df.drop(columns=["price"]),new_df["price"],test_size=0.2,random_state=42)
```

```python
df["city"].value_counts()
```

```python
encoded_col=["city"]
```

```python
trf1=ColumnTransformer(transformers=[
    ('ohe',OneHotEncoder(drop="first",sparse_output=False,dtype=np.int32,handle_unknown='ignore'),encoded_col)
],remainder='passthrough')
```

```python
trf1.fit_transform(x_train)
trf1.transform(x_test)
```

```python
x_train.head()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
trf2=ColumnTransformer(transformers=[
    ('std',StandardScaler(),slice(0,5))
],remainder='passthrough')
```

```python
trf2.fit_transform(x_train)
trf2.transform(x_test)
```

MAKE THE PIPELINE USE FIRST LINEAR REGGRESSION


```python
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LinearRegression
```

```python
pipe1=make_pipeline(trf1,trf2,LinearRegression())
```

```python
pipe1.fit(x_train,y_train)
```

```python
y_predict=pipe1.predict(x_test)
```

Import rmse mse mae for accuracy measurement

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
print("Mean_Absolute_Error :", mean_absolute_error(y_test,y_predict))
print("Mean_Squared_Error: " ,mean_squared_error(y_test,y_predict))
print("R2_Score :", r2_score(y_test,y_predict))
```
